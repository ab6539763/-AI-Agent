## 1. 第一步：运行环境自检

## 2. 第二步：四大基础数据类型与大模型参数映射
Python 中最核心的 4 种基础数据类型，在未来的大模型开发中每天都会用到：
- `str`（字符串）：模型名、Prompt 提示词、用户输入
- `int`（整数）：`max_tokens`、工龄、重试次数
- `float`（浮点数）：`temperature`（温度）、API 费用
- `bool`（布尔值）：`stream`（是否流式输出）、权限开关

```python
# 定义智枢平台员工的基础属性（演示四大基本类型）
engineer_name = "陈默"                  # str: 字符串，用双引号或单引号包裹
role_title = "AI 应用开发工程师"         # str: 岗位名称
years_of_exp = 2                         # int: 整数，不能带小数点
api_budget_per_hour = 12.5               # float: 浮点数，用于金额或温度参数
has_llm_permission = True                # bool: 布尔值，只有 True 和 False 两个值

# 使用 type() 查看每个变量在内存中的真实类型
print("姓名类型:", type(engineer_name))
print("工龄类型:", type(years_of_exp))
print("预算类型:", type(api_budget_per_hour))
print("权限类型:", type(has_llm_permission))
```

## 3. 第三步：类型转换与名片渲染实战
在实际交互中，`input()` 拿到的永远是字符串 `str`（例如 `"2"` 和 `"12.5"`），必须通过 `int()` 和 `float()` 显式转换后才能参与数学计算或格式化。

```python
# 模拟从命令行 input() 拿到的原始字符串输入
raw_years = "2"
raw_budget = "12.5"
raw_perm = "yes"

# 显式类型转换
years_exp = int(raw_years)
hourly_cost = float(raw_budget)
llm_enabled = raw_perm.strip().lower() in ("y", "yes", "是")

# 组装企业数字名片
status_label = "已开通 [OK]" if llm_enabled else "未开通 [LOCKED]"
border = "=" * 46

card_lines = [
    border,
    "   星辰数科 · 智枢平台数字名片 (v0.1.0-day01)",
    border,
    f"姓名 / 标识 : {engineer_name}",
    f"所属岗位    : {role_title}",
    f"工程经验    : {years_exp} 年",
    f"资源预算    : ¥{hourly_cost:.2f} / 小时",
    f"大模型权限  : {status_label}",
    border,
]

print("\n".join(card_lines))
```

## 4. 第四步：API 费用估算小练习（课后作业预览）
假设 DeepSeek API 输入 1 元/百万 Token，输出 2 元/百万 Token，算一算今天测试消耗 50,000 输入 + 20,000 输出要花多少钱？

```python
input_tokens = 50_000   # Python 支持用下划线分隔大数字，提高可读性
output_tokens = 20_000

total_cost = (input_tokens / 1_000_000) * 1.0 + (output_tokens / 1_000_000) * 2.0
print(f"本次测试预估花费: ¥{total_cost:.4f} 元（不到一毛钱！）")
```
